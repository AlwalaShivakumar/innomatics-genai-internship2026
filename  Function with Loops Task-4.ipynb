Problem 1 – Mobile Recharge Validation

```python
def recharge_check():
    available_plans = [199, 299, 399, 599]

    while True:
        amt = float(input("Enter recharge amount: "))
        chosen_plan = int(input("Select plan (199/299/399/599): "))

        if amt < 50:
            print("Recharge failed. Minimum amount is 50.")
        elif chosen_plan not in available_plans:
            print("Plan does not exist.")
        else:
            print("Recharge completed successfully.")
            break

        again = input("Try again? (yes/no): ").lower()
        if again != "yes":
            break


recharge_check()
```

Problem 2 – Inventory Reorder Alert

```python
def inventory_status(stock_data):
    for product in stock_data:
        quantity = stock_data[product]

        if quantity < 15:
            print(product, "needs reorder.")
        else:
            print(product, "stock level is sufficient.")


store_items = {
    "Milk": 12,
    "Biscuits": 30,
    "Toothpaste": 8,
    "Shampoo": 18
}

inventory_status(store_items)
```

Problem 3 – Student Result Processing

```python
def evaluate_marks(mark_list):
    sum_marks = 0

    for value in mark_list:
        sum_marks = sum_marks + value

    avg = sum_marks / len(mark_list)

    return "Pass" if avg >= 50 else "Fail"


scores = [72, 40, 55, 63]
print("Final Result:", evaluate_marks(scores))
```

Problem 4 – Cab Fare Estimator

```python
def fare_calculator(km_travelled, peak_time):
    fixed_charge = 50
    total_cost = fixed_charge + (km_travelled * 12)

    if peak_time == "yes":
        total_cost = total_cost * 1.25

    return total_cost


while True:
    distance_input = float(input("Distance travelled (km): "))
    peak_input = input("Peak time? (yes/no): ").lower()

    bill = fare_calculator(distance_input, peak_input)
    print("Estimated Fare:", bill)

    repeat = input("Calculate again? (yes/no): ").lower()
    if repeat != "yes":
        break
```

Problem 5 – Attendance Eligibility

```python
def attendance_check(records):
    present_days = 0

    for status in records:
        if status == "P":
            present_days += 1

    percent = (present_days / len(records)) * 100

    if percent >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


attendance_data = ["P", "A", "P", "P", "P", "A", "P", "P"]
print(attendance_check(attendance_data))
```

Problem 6 – Password Strength Checker

```python
def password_validator(pwd):
    if len(pwd) < 8:
        return "Password too short"

    digit_flag = False
    special_flag = False

    for character in pwd:
        if character.isnumeric():
            digit_flag = True
        if character in "@#$":
            special_flag = True

    if digit_flag and special_flag:
        return "Strong Password"
    else:
        return "Weak Password"


user_password = input("Enter password: ")
print(password_validator(user_password))
```
