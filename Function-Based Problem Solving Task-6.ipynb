Problem 1: Smart Parking Lot Management System

```python
def smart_parking(capacity, logs):
    current_vehicles = 0
    peak_usage = 0
    
    for log in logs:
        if log == "IN":
            current_vehicles += 1
        elif log == "OUT" and current_vehicles > 0:
            current_vehicles -= 1
        
        # Track peak usage
        if current_vehicles > peak_usage:
            peak_usage = current_vehicles
    
    # Check capacity
    if current_vehicles > capacity:
        status = "Exceeded Capacity"
    else:
        status = "Available"
    
    print("Currently Parked Vehicles:", current_vehicles)
    print("Peak Usage:", peak_usage)
    print("Parking Status:", status)


# Sample Test
smart_parking(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

Problem 2: Online Food Delivery Time Estimator

```python
def estimate_delivery_time(distance, traffic, weather):
    base_time = distance * 5
    
    # Traffic delay
    if traffic.lower() == "high":
        base_time += 15
    elif traffic.lower() == "medium":
        base_time += 10
    
    # Weather delay
    if weather.lower() == "rainy":
        base_time += 10
    elif weather.lower() == "storm":
        base_time += 20
    
    print("Estimated Delivery Time:", base_time, "minutes")


# Sample Test
estimate_delivery_time(8, "High", "Rainy")
```

Problem 3: Movie Theatre Seat Occupancy Analyzer

```python
def seat_occupancy(total_seats, booked_seats):
    booked_count = len(booked_seats)
    occupancy = (booked_count / total_seats) * 100
    
    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Seats Available"
    
    print("Occupancy:", round(occupancy, 2), "%")
    print("Show Status:", status)


# Sample Test
seat_occupancy(200, [1]*150)
```

Problem 4: Cloud Server Load Classification

```python
def server_load(cpu_readings):
    total = 0
    
    for value in cpu_readings:
        total += value
    
    average = total / len(cpu_readings)
    
    if average < 50:
        status = "Normal"
    elif 50 <= average <= 80:
        status = "Warning"
    else:
        status = "Critical"
    
    print("Average CPU Load:", round(average, 2), "%")
    print("Server Status:", status)


# Sample Test
server_load([45, 60, 70, 85, 90])
```

Problem 5: Smart Classroom Resource Usage Monitor

```python
def classroom_monitor(resource_usage):
    overused = []
    
    for resource, hours in resource_usage.items():
        if hours > 8:
            overused.append(resource)
    
    if overused:
        alert = "Yes"
    else:
        alert = "No"
    
    print("Overused Resources:", ", ".join(overused) if overused else "None")
    print("Energy Alert:", alert)


# Sample Test
classroom_monitor({
    "Projector": 6,
    "AC": 9,
    "Lights": 4
})
```

Problem 6: Online Event Registration Capacity Controller

```python
def event_registration(capacity, registrations):
    if registrations > capacity:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"
    else:
        confirmed = registrations
        waitlisted = 0
        status = "Open"
    
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)


# Sample Test
event_registration(100, 105)
```
