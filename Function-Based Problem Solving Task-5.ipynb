Problem 1: Social Media – Post Engagement Analyzer

```python
def analyze_post_engagement(likes_list):
    total_likes = 0
    
    for likes in likes_list:
        total_likes += likes
    
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    
    return total_likes, status


# Example
likes_data = [200, 150, 300, 400]
total, result = analyze_post_engagement(likes_data)

print("Total Likes:", total)
print("Post Status:", result)
```

Problem 3: Agriculture – Rainfall Adequacy Checker

```python
def check_rainfall(rainfall_list, required_level):
    total = 0
    
    for value in rainfall_list:
        total += value
    
    average = total / len(rainfall_list)
    
    if average >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"
    
    return average, status


# Example
rain_data = [70, 75, 68, 75]
avg, status = check_rainfall(rain_data, 70)

print("Average Rainfall:", avg)
print("Rainfall Status:", status)
```

Problem 4: Social Media – Duplicate Account Detection

```python
def detect_duplicate_accounts(usernames):
    unique_users = set(usernames)
    
    if len(unique_users) < len(usernames):
        result = "Yes"
    else:
        result = "No"
    
    return result


# Example
users = ["john", "alice", "john", "mark"]
print("Duplicate Accounts Found:", detect_duplicate_accounts(users))
```

Problem 6: Agriculture – Premium Crop Price Filter

```python
def filter_premium_crops(price_list):
    premium = []
    
    for price in price_list:
        if price > 2000:
            premium.append(price)
    
    return premium


# Example
prices = [1500, 2500, 3200, 1800]
premium_list = filter_premium_crops(prices)

print("Premium Crops:", premium_list)
```

Problem 7: System Monitoring – Application Health Checker

```python
def check_application_health(error_count):
    if error_count == 0:
        status = "Healthy"
    elif error_count <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"
    
    return error_count, status


# Example
errors, status = check_application_health(7)
print("Error Count:", errors)
print("System Status:", status)
```

Problem 8: Banking – Daily Transaction Limit Checker

```python
def check_transaction_limit(amount):
    daily_limit = 50000
    
    if amount <= daily_limit:
        status = "Approved"
    else:
        status = "Rejected"
    
    return amount, status


# Example
amount, status = check_transaction_limit(60000)
print("Transaction Amount:", amount)
print("Transaction Status:", status)
```

Problem 9: E-Learning – Student Attendance Eligibility System

```python
def check_exam_eligibility(attendance_list):
    total_classes = len(attendance_list)
    attended = 0
    
    for status in attendance_list:
        if status == 1:
            attended += 1
    
    percentage = (attended / total_classes) * 100
    
    if percentage >= 75:
        eligibility = "Eligible"
    else:
        eligibility = "Not Eligible"
    
    return percentage, eligibility


# Example (1 = Present, 0 = Absent)
attendance = [1, 1, 0, 1, 1]
percent, status = check_exam_eligibility(attendance)

print("Attendance Percentage:", percent)
print("Exam Eligibility:", status)
```

Problem 10: Smart Electricity Bill Analyzer

```python
def calculate_electricity_bill(units):
    bill = 0
    
    # Slab calculation
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)
    
    # Usage classification
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"
    
    return bill, usage


# Example
bill_amount, usage_status = calculate_electricity_bill(250)

print("Total Bill:", bill_amount)
print("Usage Classification:", usage_status)
```
